**Python - Sintaxis Básica:**

Escribe un programa en Python que tome dos números ingresados por el usuario y muestre la suma de ambos.


```python
numero_1 = float(input("Ingresa un número: "))
numero_2 = float(input("Ingresa otro número: "))
suma = numero_1 + numero_2
print("La suma de", numero_1, "+", numero_2, "es", suma)
```

Python - Estructuras de Control:

Desarrolla un programa que determine si un número ingresado por el usuario es par o impar.


```python
numero = float(input("Ingresa un número para saber si es par o impar: "))
if numero % 2 == 0:
  print("El número es par")
else:
  print("El número es impar")
```

Implementa un programa que solicite al usuario su edad y, según la edad, muestre un mensaje indicando si es menor de edad o mayor de edad.


```python
user_age = int(input("Ingresa tu edad: "))
if user_age < 18:
  print("El usuario es menor de edad")
else:
  print("El usuario es mayor de edad, puede proceder")
```

Python - Manipulación de Datos:


Crea una lista de nombres de ciudades. Luego, utiliza Python para mostrar la longitud de la lista y ordenarla alfabéticamente.


```python
cities_list = ["Tokio", "Shanghai", "Nueva Dheli", "Sao Paulo", "CDMX", "Bombay", "Amsterdam"]
longitud = len(cities_list)
print("La longitud de la lista es de", longitud, "elementos")
```

```python
sorted_list = sorted(cities_list)
sorted_list
```

Utiliza la librería NumPy para realizar una operación matemática simple, como la suma o la multiplicación de dos arreglos NumPy.


```python
import numpy as np

arreglo1 = np.array([1,3,5,7,9])
arreglo2 = np.array([2,4,6,8,10])

operacion = arreglo1 * arreglo2
print("El producto obtenido de multiplicar los 2 arreglos es: ", operacion)
```
